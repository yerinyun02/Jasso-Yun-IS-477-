## Data exploration and Plot

The EIA-860 dataset describes supply-side data, which provides information on power plant capacity, producer type, and fuel sources such as coal, natural gas, wind, and solar. the EIA-861 dataset represents demand-side, which includes electricity sales (MWh) and average prices (cents per kWh) across sectors and states. 

For Dataset EIA 860, using OpenRefine, I filtered rows to 2014–2324 only from the Year column and excluded "All Sources" rows to avoid double-counting. I exported it as CSV format. For Dataset EIA-861, I will use the ‘Total Electric Industry’ sheet from the EIA-861 dataset to limit the dataset scope. I will parse 3-line header columns into one row of columns to reduce the complexity of the dataset 
pe.

```python
# EIA_860
# Turning the first row into column headers
import pandas as pd

eia_860 = pd.read_csv('EIA-860.csv', header = 1)
print(eia_860.columns)
# first_row = df.iloc[0].tolist()

eia_860.to_csv('EIA_860_clean.csv', index = False)
```

```python
# EIA_861 
# Read the file and transform to a CSV file
eia_861 = pd.read_excel("EIA 861_Total Electric Industry.xlsx", header = [0,1,2], index_col = None)

new_cols = []
for i, col in enumerate(eia_861.columns):
    level0, level1, level2 = col
    if i == 0:
        new_cols.append('Year')
    elif i == 1:
        new_cols.append('STATE')
    else:
        new_cols.append(f"{level0} - {level1} ({level2})")

eia_861.columns = new_cols

eia_861.to_csv('EIA_861_Total_Electric_Industry_clean.csv', index=False)
```

```python
eia_861.head()
```

```python
#: :generated:
#: :model: Claude Sonnet 4.6
#: :prompt: Merge two cleaned dataset based on Year and State Code. 
#: :changes: In the merged dataset, make sure it has fuel type.
#: :response:
# --- Filter EIA-860 to 2014-2024 only ---
eia_860 = eia_860[eia_860['Year'].between(2014, 2024)]

# --- Merge on Year + State (keep all EIA-860 rows including Producer Type & Fuel Source) ---
merged = eia_860.merge(
    eia_861,
    left_on=['Year', 'State Code'],   # EIA-860 key columns
    right_on=['Year', 'STATE'],        # EIA-861 key columns
    how='inner'                        # only keep rows that exist in BOTH datasets
)

# --- Drop the duplicate state column ---
merged = merged.drop(columns='STATE')

#  --- Remove double-counting summary rows ---
merged = merged[merged['Producer Type'] != 'Total Electric Power Industry']

# --- Save ---
merged.to_csv('EIA_merged_final.csv', index=False)
#: :end:
```

The merged dataset has one row per state, year, producer type, and fuel source, which is 10,800 rows. While this detail is useful, it is too granular for building an interactive dashboard. Therefore, I am creating a simpler summary tables that has one row per state per year, with capacity shares of renewable, fossil, and IPP as well as different sectors' electricity prices.

```python
#: :generated:
#: :model: Claude Sonnet 4.6
#: :prompt: Process 'EIA_merged_final.csv' based on fuel types (renewable vs fossil fuels). Then, make sure it has state, year, and fuel source per each row. Having capacity shares of renewable, fossil, and IPP calculated at the state-year level. Lastly, include electricity prices (residential, commercial, industrial) attached to each row
#: :response:
# Define fuel categories
renewable_fuels = [
    'Wind', 'Solar Thermal and Photovoltaic', 'Hydroelectric',
    'Geothermal', 'Wood and Wood Derived Fuels', 'Other Biomass'
]
fossil_fuels = ['Coal', 'Natural Gas', 'Petroleum', 'Other Gases']

# Group by state and year
group = merged.groupby(['Year', 'State Code'])

# Calculate capacity totals
total_cap  = group['Nameplate Capacity (Megawatts)'].sum()
renew_cap  = (merged[merged['Fuel Source'].isin(renewable_fuels)]
              .groupby(['Year', 'State Code'])['Nameplate Capacity (Megawatts)'].sum())
fossil_cap = (merged[merged['Fuel Source'].isin(fossil_fuels)]
              .groupby(['Year', 'State Code'])['Nameplate Capacity (Megawatts)'].sum())
ipp_cap    = (merged[merged['Producer Type'] == 'Electric Generators, Independent Power Producers']
              .groupby(['Year', 'State Code'])['Nameplate Capacity (Megawatts)'].sum())

# Build summary (temporary — only needed to calculate shares)
summary = pd.DataFrame({
    'total_cap':  total_cap,
    'renew_cap':  renew_cap,
    'fossil_cap': fossil_cap,
    'ipp_cap':    ipp_cap,
    'res_price':  group['RESIDENTIAL - Price (Cents/kWh)'].first(),
    'com_price':  group['COMMERCIAL - Price (Cents/kWh)'].first(),
    'ind_price':  group['INDUSTRIAL - Price (Cents/kWh)'].first(),
}).reset_index()

summary['renew_share']  = (summary['renew_cap']  / summary['total_cap'] * 100).round(1)
summary['fossil_share'] = (summary['fossil_cap'] / summary['total_cap'] * 100).round(1)
summary['ipp_share']    = (summary['ipp_cap']    / summary['total_cap'] * 100).round(1)
summary = summary.dropna(subset=['renew_share', 'res_price'])

# Build fuel_merged — this is the only file we need to save
fuel_cap = (
    merged[merged['Fuel Source'] != 'All Sources']
    .groupby(['Year', 'State Code', 'Fuel Source'])['Nameplate Capacity (Megawatts)']
    .sum()
    .reset_index()
)

fuel_merged = fuel_cap.merge(
    summary[['Year', 'State Code', 'renew_share', 'ipp_share',
             'fossil_share', 'res_price', 'com_price', 'ind_price']],
    on=['Year', 'State Code'],
    how='left'
)

# Save only fuel_merged — no need for summary.csv
fuel_merged.to_csv('fuel_merged.csv', index=False)

fuel_merged.head()
#: :end:
```

# Average Residential Price by Year (line plot)

```python
import matplotlib.pyplot as plt

avg_price_year = merged.groupby('Year')['RESIDENTIAL - Price (Cents/kWh)'].mean()
avg_price_year.plot()
plt.title('Average Residential Electricity Price by Year')
plt.xlabel('Year')
plt.ylabel('Price (Cents/kWh)')
plt.show()
```
